## 4. Inspection de la nature et des modalités des variables

Bien que toutes les variables soient encodées numériquement, leur type informatique ne permet pas à lui seul de déterminer leur nature statistique. Nous examinons donc le nombre de valeurs distinctes ainsi que les modalités des variables afin de distinguer notamment les variables quantitatives des variables catégorielles codées numériquement.

```python
DF.nunique().sort_values()
```

La méthode `nunique()` a permis d’identifier le nombre de valeurs distinctes présentes dans chaque variable. Les résultats montrent que certaines variables possèdent très peu de modalités : `sex`, `fbs`, `target` et `exang` n’en comptent que 2 ; `slope`, `thal` et `restecg` en comptent 3 ; tandis que `cp` et `ca` en comptent 4. Bien que ces variables soient encodées numériquement, leur faible nombre de modalités indique que plusieurs d’entre elles représentent des variables catégorielles codées sous forme numérique. À l’inverse, `oldpeak`, `age`, `trestbps`, `thalach` et `chol` présentent respectivement 40, 41, 50, 91 et 152 valeurs distinctes, ce qui est cohérent avec leur caractère quantitatif. Cette inspection permet ainsi de mieux distinguer la nature statistique des variables au-delà de leur simple type informatique (`int64` ou `float64`).

```python
# Affichage des modalités des variables ayant peu de valeurs distinctes
for col in DF.columns:
    if DF[col].nunique() <= 10:
        modalites = [int(x) for x in sorted(DF[col].unique())]
        print(f"{col} : {modalites}")
```

**Commentaire** : L'inspection des modalités confirme que certaines variables, bien qu'elles soient enregistrées sous un type numérique (int64), représentent en réalité des caractéristiques catégorielles codées numériquement. Les variables `sex`, `fbs`, `exang` et `target` sont binaires, tandis que `cp`, `restecg`, `slope` et `thal` possèdent plusieurs modalités codées par des nombres entiers. La variable `ca` prend les valeurs 0 à 3 et correspond à un nombre discret. Cette distinction entre type informatique et nature statistique des variables sera prise en compte dans les analyses ultérieures. Les principales variables quantitatives sont `age`, `trestbps`, `chol`, `thalach` et `oldpeak`.

### 5.1 Distribution des variables catégorielles

Nous examinons les effectifs et les proportions associés à chaque modalité.

```python
var_categorielles=['sex', 'cp', 'fbs', 'restecg','exang', 'slope', 'thal', 'target']

for var in var_categorielles:
    
    effectifs = DF[var].value_counts().sort_index()
    pourcentages = (DF[var].value_counts(normalize=True).sort_index().mul(100).round(2))
    resume=pd.DataFrame({'Effectifs':effectifs,'Pourcentage (%)':pourcentages})

    print(f"\nVariable : {var}")
    display(resume)
```

**Commentaire :**
L’analyse des effectifs et des pourcentages permet d’examiner la répartition des observations entre les différentes modalités des variables catégorielles. Pour la variable *"sex"*, la modalité 1 est majoritaire avec 206 observations (67,99 %) contre 97 (32,01 %) pour la modalité 0. La variable *"cp"* est principalement représentée par la modalité 3 avec 144 observations (47,52 %), suivie des modalités 2 (28,38 %), 1 (16,50 %) et 0 (7,59 %). Pour *"fbs"*, la modalité 0 est très largement majoritaire avec 258 observations (85,15 %). Concernant *"restecg"*, les modalités 0 (49,83 %) et 2 (48,84 %) sont presque équitablement représentées, tandis que la modalité 1 est très peu fréquente (1,32 %). Pour *"exang"*, la modalité 0 représente 67,33 % des observations contre 32,67 % pour la modalité 1. La variable *"slope"* présente une répartition presque identique entre les modalités 0 (46,86 %) et 1 (46,20 %), alors que la modalité 2 ne représente que 6,93 % des observations. Pour *"thal"*, la modalité 1 est majoritaire avec 55,45 %, suivie de la modalité 3 avec 38,61 %, tandis que la modalité 2 est minoritaire (5,94 %). Enfin, la variable cible target présente une répartition relativement équilibrée : 164 observations (54,13 %) pour la modalité 0 contre 139 observations (45,87 %) pour la modalité 1. Dans l’ensemble, certaines variables présentent donc des modalités fortement déséquilibrées, notamment *"fbs"*, tandis que la variable cible *"target"* reste relativement équilibrée.
