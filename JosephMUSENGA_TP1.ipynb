```python
# Import des bibliothèques de travail
import pandas as pd
import numpy as np
```

```python
#Import du jeu de données à utiliser:

#Définir l'URL
# Source publique du jeu de données via mon repository github
url = "https://raw.githubusercontent.com/jomusenga/data_preprocessing_tp/main/data/dataset_anomalies.csv"

# Importation des données
source_data = pd.read_csv(url)

# Aperçu des données
source_data.head()
```

```python
# On crée une copie du jeu de données pour éviter des modifications dans le fichier source pendant la manipulation.
DF=source_data.copy()
```

```python
# On verifie le type des variables et autres informations générales
DF.info()
```

```python
#Convertion des variables "Revenu" et "IMC" en numérique
DF['Revenu']=pd.to_numeric(DF['Revenu'],errors='coerce')
DF['IMC']=pd.to_numeric(DF['IMC'],errors='coerce')
```

```python
# On Vérifie le nombre de valeurs manquantes par variable
DF.isna().sum()
```

```python
# On evalue en pourcentage le nombre des valeurs manquantes par variable
round((DF.isnull().sum()/len(DF))*100,2).astype(str) + "%"
```

```python
# On compte le nombre de modalités "unknown" par variable
unknown_count = (DF == 'unknown').sum()

# On Conserve uniquement les variables contenant des valeurs "unknown"
unknown_count = unknown_count[unknown_count > 0]

# On Calcule le pourcentage de valeurs "unknown"
unknown_percent = round((unknown_count / len(DF) * 100),2)

# On Présente les résultats dans un tableau
unknown_summary = pd.DataFrame({'Nombre_unknown': unknown_count,'Pourcentage_unknown (%)': unknown_percent})

unknown_summary
```

```python
# On compte les observations dupliquées
print(DF.duplicated().sum())
```

```python
# On affiche le nombre des modalités pour chaque variable catégorielle
for colonne in DF.select_dtypes(include="str").columns:
    print(colonne,":", DF[colonne].nunique(),"modalités")
```

```python
# On affiche la frequence des modalités de chaque variable catégorielle.
for colonne in DF.select_dtypes(include="str").columns:
    print("\nVariable :",colonne)
    print(DF[colonne].value_counts())
```

```python
# On affiche la frequence (en %) des modalités de chaque variable catégorielle
for colonne in DF.select_dtypes(include="str").columns:
    print("\nVariable:",colonne)
    print(DF[colonne].value_counts(normalize=True)*100)
```

```python
for colonne in DF.select_dtypes(include="str").columns:
    print("\nVariable:",colonne)
    print((DF[colonne].value_counts(normalize=True)*100).round(2))
```
